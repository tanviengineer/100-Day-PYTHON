10. Text Classification:
 Build a text classifier (e.g. sentiment analysis, spam detection) using machine
learning algorithms (e.g. Naive Bayes, SVM, neural networks)
 Train and evaluate the classifier on labeled datasets.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
import requests
import zipfile
import io
```

```python
# load url
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00228/smsspamcollection.zip"
```

```python
# download zip file
response = requests.get(url)
zip_file = zipfile.ZipFile(io.BytesIO(response.content))
```

```python
with zip_file.open("SMSSpamCollection") as file:
  df=pd.read_csv(file,sep="\t",names=["label","message"])
print(df.head())
```

```python
# split dataset
X_train, X_test, y_train ,y_test = train_test_split(
    df["message"],
    df["label"],
    test_size=0.2,
    random_state=42
)

```

```python
# convert text into numerical tf-idf features
tfidf = TfidfVectorizer(stop_words="english")

X_train = tfidf.fit_transform(X_train)
X_test = tfidf.transform(X_test)
```

```python
# train naive  bayes classifer
model = MultinomialNB()
model.fit(X_train, y_train)
```

```python
# test classifier
predictions= model.predict(X_test)
```

```python
# evaluation
print("Accuracy", accuracy_score(y_test,predictions))
print(classification_report(y_test,predictions))
```

```python
# test new message
message=["congratulations ! you won a free mobile phone .click here now !"]
message = tfidf.transform(message)

print("prediction: ", model.predict(message)[0])
```
